```python
# Check how many of the equal scores are integers
equal_scores_mask = df['score_me'] == df['score_other']
equal_scores = df.loc[equal_scores_mask, 'score_me']
integer_equal_scores_count = equal_scores.apply(lambda x: float(x).is_integer()).sum()
integer_equal_scores_count
```

# Ellipse

## Raw

```python
ellipse_df = pd.read_csv(r"C:\Users\alvar\Githubs\mentor-eval\data\raw\ellipse\ELLIPSE_Final_github_train.csv")
ellipse_df.shape
```

```python
ellipse_df.head()
```

```python
# Create total_grade column as sum of the 6 scoring dimensions, rounded up to nearest integer
ellipse_df['total_grade'] = np.ceil(ellipse_df['Cohesion'] + 
                                   ellipse_df['Syntax'] + 
                                   ellipse_df['Vocabulary'] + 
                                   ellipse_df['Phraseology'] + 
                                   ellipse_df['Grammar'] + 
                                   ellipse_df['Conventions']).astype(int)

# Calculate min and max values for each scoring dimension
scoring_columns = ['Cohesion', 'Syntax', 'Vocabulary', 'Phraseology', 'Grammar', 'Conventions']
print("Min values for each scoring dimension:")
print(ellipse_df[scoring_columns].min())
print("\nMax values for each scoring dimension:")
print(ellipse_df[scoring_columns].max())

# Calculate min and max for the aggregated total_grade
print(f"\nTotal grade - Min: {ellipse_df['total_grade'].min()}, Max: {ellipse_df['total_grade'].max()}")
```
